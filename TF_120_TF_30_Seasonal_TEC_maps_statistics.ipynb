# TF-30 and TF-120 TEC maps

```python
import numpy as np
import os
from pathlib import Path

def process_tec_maps(resolution, source):
    print(f"Processing {resolution} {source} TEC maps...")

    file_patterns = [
        f"{resolution}_{source}_TEC_maps_0800.npy",
        f"{resolution}_{source}_TEC_maps_1600.npy",
        f"{resolution}_{source}_TEC_maps_2000_0400.npy"
    ]

    files_to_process = []
    for pattern in file_patterns:
        if os.path.exists(pattern):
            files_to_process.append(pattern)
        else:
            print(f"Warning: File {pattern} not found.")

    if not files_to_process:
        print(f"No files found for {resolution} {source}")
        return

    print(f"Files to process in order: {files_to_process}")

    arrays = []
    for file in files_to_process:
        data = np.load(file)
        print(f"Loaded {file} with shape: {data.shape}")
        arrays.append(data)

    concatenated = np.concatenate(arrays, axis=0)
    print(f"Concatenated shape: {concatenated.shape}")

    output_file = f"{resolution}_{source}_TEC_maps_combined.npy"
    np.save(output_file, concatenated)
    print(f"Saved concatenated data to {output_file}")

    total_elements = concatenated.size
    negative_values = np.sum(concatenated < 0)
    high_values = np.sum(concatenated > 150)
    zero_values = np.sum(concatenated == 0)
    nan_values = np.sum(np.isnan(concatenated))

    negative_percent = (negative_values / total_elements) * 100
    high_percent = (high_values / total_elements) * 100
    zero_percent = (zero_values / total_elements) * 100
    nan_percent = (nan_values / total_elements) * 100

    print(f"Statistics for {resolution} {source}:")
    print(f"  Total elements: {total_elements}")
    print(f"  Negative values: {negative_values} ({negative_percent:.4f}%)")
    print(f"  Values > 150: {high_values} ({high_percent:.4f}%)")
    print(f"  Zero values: {zero_values} ({zero_percent:.4f}%)")
    print(f"  NaN values: {nan_values} ({nan_percent:.4f}%)")
    print("-" * 50)

    return {
        "resolution": resolution,
        "source": source,
        "total_elements": total_elements,
        "negative_values": negative_values,
        "negative_percent": negative_percent,
        "high_values": high_values,
        "high_percent": high_percent,
        "zero_values": zero_values,
        "zero_percent": zero_percent,
        "nan_values": nan_values,
        "nan_percent": nan_percent
    }

def main():

    resolutions = ["TF_30", "TF_120"]

    sources_30 = ["EMBRACE", "MAGGIA", "Nagoya"]
    sources_120 = ["EMBRACE", "IGS", "MAGGIA", "Nagoya"]

    results = []

    for source in sources_30:
        result = process_tec_maps("TF_30", source)
        if result:
            results.append(result)

    for source in sources_120:
        result = process_tec_maps("TF_120", source)
        if result:
            results.append(result)

    print("\nSummary Report:")
    print("-" * 100)
    headers = ["Resolution", "Source", "Total Elements", "Negative %", "Values > 150 %", "Zero %", "NaN %"]
    print(f"{headers[0]:<10} {headers[1]:<10} {headers[2]:<15} {headers[3]:<12} {headers[4]:<15} {headers[5]:<10} {headers[6]:<10}")
    print("-" * 100)

    for result in results:
        print(f"{result['resolution']:<10} {result['source']:<10} "
              f"{result['total_elements']:<15} {result['negative_percent']:.4f}%{'':<5} "
              f"{result['high_percent']:.4f}%{'':<8} {result['zero_percent']:.4f}%{'':<3} "
              f"{result['nan_percent']:.4f}%")

    try:
        import csv
        with open('tec_maps_statistics.csv', 'w', newline='') as csvfile:
            fieldnames = ['Resolution', 'Source', 'Total Elements',
                         'Negative Values', 'Negative %',
                         'Values > 150', 'Values > 150 %',
                         'Zero Values', 'Zero %',
                         'NaN Values', 'NaN %']
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

            writer.writeheader()
            for result in results:
                writer.writerow({
                    'Resolution': result['resolution'],
                    'Source': result['source'],
                    'Total Elements': result['total_elements'],
                    'Negative Values': result['negative_values'],
                    'Negative %': f"{result['negative_percent']:.4f}%",
                    'Values > 150': result['high_values'],
                    'Values > 150 %': f"{result['high_percent']:.4f}%",
                    'Zero Values': result['zero_values'],
                    'Zero %': f"{result['zero_percent']:.4f}%",
                    'NaN Values': result['nan_values'],
                    'NaN %': f"{result['nan_percent']:.4f}%"
                })
        print(f"\nStatistics also saved to tec_maps_statistics.csv")
    except Exception as e:
        print(f"Could not save CSV file: {e}")

if __name__ == "__main__":
    main()
```

# Seasonal TEC maps

```python
import numpy as np
import os
from pathlib import Path

def process_seasonal_tec_maps(source):
    print(f"Processing Seasonal {source} TEC maps...")

    file_name = f"Seasonal_{source}_TEC_maps.npy"

    if not os.path.exists(file_name):
        print(f"Warning: File {file_name} not found.")
        return None

    print(f"Processing file: {file_name}")

    data = np.load(file_name)
    print(f"Loaded {file_name} with shape: {data.shape}")

    total_elements = data.size
    negative_values = np.sum(data < 0)
    high_values = np.sum(data > 150)
    zero_values = np.sum(data == 0)
    nan_values = np.sum(np.isnan(data))

    negative_percent = (negative_values / total_elements) * 100
    high_percent = (high_values / total_elements) * 100
    zero_percent = (zero_values / total_elements) * 100
    nan_percent = (nan_values / total_elements) * 100

    print(f"Statistics for Seasonal {source}:")
    print(f"  Total elements: {total_elements}")
    print(f"  Negative values: {negative_values} ({negative_percent:.4f}%)")
    print(f"  Values > 150: {high_values} ({high_percent:.4f}%)")
    print(f"  Zero values: {zero_values} ({zero_percent:.4f}%)")
    print(f"  NaN values: {nan_values} ({nan_percent:.4f}%)")
    print("-" * 50)

    return {
        "source": source,
        "total_elements": total_elements,
        "negative_values": negative_values,
        "negative_percent": negative_percent,
        "high_values": high_values,
        "high_percent": high_percent,
        "zero_values": zero_values,
        "zero_percent": zero_percent,
        "nan_values": nan_values,
        "nan_percent": nan_percent
    }

def main():

    sources = ["EMBRACE", "IGS", "MAGGIA", "Nagoya"]

    results = []

    for source in sources:
        result = process_seasonal_tec_maps(source)
        if result:
            results.append(result)

    print("\nSummary Report:")
    print("-" * 100)
    headers = ["Source", "Total Elements", "Negative %", "Values > 150 %", "Zero %", "NaN %"]
    print(f"{headers[0]:<10} {headers[1]:<15} {headers[2]:<12} {headers[3]:<15} {headers[4]:<10} {headers[5]:<10}")
    print("-" * 100)

    for result in results:
        print(f"{result['source']:<10} "
              f"{result['total_elements']:<15} {result['negative_percent']:.4f}%{'':<5} "
              f"{result['high_percent']:.4f}%{'':<8} {result['zero_percent']:.4f}%{'':<3} "
              f"{result['nan_percent']:.4f}%")

    try:
        import csv
        with open('seasonal_tec_maps_statistics.csv', 'w', newline='') as csvfile:
            fieldnames = ['Source', 'Total Elements',
                         'Negative Values', 'Negative %',
                         'Values > 150', 'Values > 150 %',
                         'Zero Values', 'Zero %',
                         'NaN Values', 'NaN %']
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

            writer.writeheader()
            for result in results:
                writer.writerow({
                    'Source': result['source'],
                    'Total Elements': result['total_elements'],
                    'Negative Values': result['negative_values'],
                    'Negative %': f"{result['negative_percent']:.4f}%",
                    'Values > 150': result['high_values'],
                    'Values > 150 %': f"{result['high_percent']:.4f}%",
                    'Zero Values': result['zero_values'],
                    'Zero %': f"{result['zero_percent']:.4f}%",
                    'NaN Values': result['nan_values'],
                    'NaN %': f"{result['nan_percent']:.4f}%"
                })
        print(f"\nStatistics also saved to seasonal_tec_maps_statistics.csv")
    except Exception as e:
        print(f"Could not save CSV file: {e}")

if __name__ == "__main__":
    main()
```
