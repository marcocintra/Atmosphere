## 2022

```python
import os
import numpy as np

current_directory = './'

files_22i = []

for root, dirs, files in os.walk(current_directory):
    for file in files:
        if file.endswith(".22i") and file.startswith("igsg"):
            files_22i.append(os.path.join(root, file))

files_22i = sorted(files_22i)
print(files_22i)

print("Files .22i found (sorted):")
for file in files_22i:
    print(file)

for file_path_i in files_22i:

    data_list = []

    with open(file_path_i, 'r') as file:
        for linha in file:
            if "START OF RMS MAP" in linha:
                print("End of TEC file found. Stopping processing.")
                break
            if "LAT/LON1/LON2/DLON/H" in linha:

                line_values1 = list(map(float, file.readline().split()))
                line_values2 = list(map(float, file.readline().split()))
                line_values3 = list(map(float, file.readline().split()))
                line_values4 = list(map(float, file.readline().split()))
                line_values5 = list(map(float, file.readline().split()))

                combined_values = line_values1 + line_values2 + line_values3 + line_values4 + line_values5

                data_list.append(combined_values)

        data_array = np.array(data_list)

        file_name_without_extension = os.path.splitext(os.path.basename(file_path_i))[0]

        output_path_npy = os.path.join(os.path.dirname(file_path_i), f"{file_name_without_extension}_22.npy")

        elements_per_line = 71

        total_lines = data_array.shape[0] // elements_per_line

        new_shape = (total_lines, elements_per_line, data_array.shape[1])
        data_array = data_array[:total_lines * elements_per_line, :].reshape(new_shape)
        print(np.shape(data_array))
        data_array = data_array[:12, 19:67+1, 14:36+1]
        print(np.shape(data_array))
        np.save(output_path_npy, data_array)

        print(f"Array saved in: {output_path_npy}")
```

```python
diretorio_principal = './'

files_npy = [file for file in os.listdir(diretorio_principal) if file.endswith('_22.npy')]
files_npy = sorted(files_npy)
print(files_npy)

if not files_npy:
    print("No .npy files found in the current directory.")
else:
    array_4d = np.empty((len(files_npy), 12, 49, 23))

    for i, file in enumerate(files_npy):
        file_path = os.path.join(diretorio_principal, file)
        array_npy = np.load(file_path)
        array_4d[i, :, :, :] = array_npy

    array_4d = array_4d/10
    np.save(os.path.join(diretorio_principal, 'IGS_TEC_maps_intersection_2022.npy'), array_4d)

    print(f"Processados {len(files_npy)} files .npy")
    print(f"Forma do array resultante: {array_4d.shape}")
```

# Dates adjusts

```python
import numpy as np
from datetime import datetime, timedelta

def doy_to_date(year, doy):
    return datetime(year, 1, 1) + timedelta(days=doy - 1)

doys = {
    2022: {
        3: [60, 61, 62, 63, 64, 65, 66, 67, 69, 70,
            71, 73, 74, 75, 76, 77, 78, 79, 80, 81,
            82, 83, 84, 85, 86, 90],
        6: [152, 153, 154, 156, 157, 158, 159, 160,
            161, 162, 163, 164, 165, 166, 167, 168,
            169, 170, 171, 172, 173, 174, 176, 178,
            181],
        9: [245, 246, 247, 248, 249, 250, 251, 252,
            253, 254, 255, 256, 258, 260, 261, 262,
            264, 265, 266, 267, 268, 269, 270, 271,
            272, 273],
        12: [338, 339, 340, 348, 349, 350, 351, 352,
             353, 354, 356, 357, 358, 359, 360, 362,
             363, 364, 365]
    },
    2024: {
        3: [61, 62, 63, 64, 65, 66, 67, 68, 69, 70,
            71, 72, 73, 74, 75, 76, 77, 78, 79, 80,
            81, 86, 87, 88, 89, 91],
        6: [153, 154, 155, 156, 157, 158, 159, 160, 161, 162,
            163, 164, 165, 166, 167, 168, 169, 170, 171, 173,
            174, 175, 176, 177, 180, 181],
        9: [247, 248, 249, 250, 251, 252, 253, 254, 255, 256,
            257, 258, 261, 262, 263, 264, 265, 266, 267, 268,
            269, 270, 271, 272, 274],
        12: [336, 337, 338, 339, 340, 341, 342, 343, 344, 345,
             346, 347, 349, 350, 351, 352, 353, 354, 355,
             356, 357, 358, 359, 360, 361, 362, 363, 364, 365]
    }
}

datetime_list = []

for year in doys.keys():
    for month in doys[year].keys():
        for doy in doys[year][month]:
            base_date = doy_to_date(year, doy)
            for hour in range(0, 24, 2):
                dt = datetime(base_date.year, base_date.month, base_date.day, hour, 0, 0)
                datetime_list.append(dt)

igs_datetimes_2022_2024 = np.array(datetime_list, dtype='datetime64[s]')

print(f"Total number of datetime points: {len(igs_datetimes_2022_2024)}")
print(f"First datetime: {igs_datetimes_2022_2024[0]}")
print(f"Last datetime: {igs_datetimes_2022_2024[-1]}")

print("\nSample of first 10 datetime entries:")
for dt in igs_datetimes_2022_2024[:10]:
    print(dt)

print("\nSample of last 10 datetime entries:")
for dt in igs_datetimes_2022_2024[-10:]:
    print(dt)
```
