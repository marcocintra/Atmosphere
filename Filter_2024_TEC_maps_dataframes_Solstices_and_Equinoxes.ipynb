## EMBRACE

```python
import pandas as pd

%cd '/content/drive/MyDrive'

df_embrace = pd.read_pickle('TF_EMBRACE_TEC_maps_2022_2024.pkl')
```

```python
df_embrace
```

```python
print("Available columns in DataFrame:")
print(df_embrace.columns.tolist())
print(f"\nDataFrame shape: {df_embrace.shape}")
```

```python
df_embrace = df_embrace.reset_index()

print("Columns after reset_index:")
print(df_embrace.columns.tolist())
```

```python
df_embrace['DATETIME'] = pd.to_datetime(df_embrace['DATETIME'])

from datetime import datetime

dates_obj = [
    datetime(2024, 3, 20, 4, 0, 0),
    datetime(2024, 6, 21, 0, 0, 0),
    datetime(2024, 9, 22, 14, 0, 0),
    datetime(2024, 12, 21, 10, 0, 0)
]

df_embrace = df_embrace[df_embrace['DATETIME'].isin(dates_obj)]
```

```python
df_embrace
```

```python
import numpy as np
```

```python
mapas_embrace = np.array(df_embrace.iloc[:]['TECMAP'])
```

```python
np.shape(df_embrace)
```

```python
np_mapas_embrace = []
for i in range(len(mapas_embrace)):
    np_mapas_embrace.append(mapas_embrace[i])
np_mapas_embrace = np.array(np_mapas_embrace)
```

```python
np.shape(np_mapas_embrace)
```

```python
type(np_mapas_embrace)
```

```python
np.save('EMBRACE_TEC_maps.npy', np_mapas_embrace)
```

```python
!ls -lh EMBRACE_maps*
```

```python
df_embrace['DATETIME'].values
```

```python
np.shape(df_embrace['DATETIME'].values)
```

```python
np.save('EMBRACE_TEC_maps_datetimes.npy', df_embrace['DATETIME'].values)
```

## IGS

```python
import pandas as pd

%cd '/content/drive/MyDrive'

df_igs = pd.read_pickle('TF_IGS_TEC_maps_2022_2024.pkl')
```

```python
df_igs
```

```python
reference_date = pd.Timestamp('2024-06-20 22:00:00')

subsequent_records = df_igs[df_igs['DATETIME'] > reference_date]

if len(subsequent_records) > 0:

    closest_record = subsequent_records.loc[subsequent_records['DATETIME'].idxmin()]

    print(f"Reference date: {reference_date}")
    print(f"Next record found: {closest_record['DATETIME']}")
    print(f"Difference: {closest_record['DATETIME'] - reference_date}")
    print("\nComplete record:")
    print(closest_record)
else:
    print("No records found after the specified date")
```

```python
df_igs['DATETIME'] = pd.to_datetime(df_igs['DATETIME'])

from datetime import datetime

dates_obj = [
    datetime(2024, 3, 20, 4, 0, 0),
    datetime(2024, 6, 21, 0, 0, 0),
    datetime(2024, 9, 22, 14, 0, 0),
    datetime(2024, 12, 21, 10, 0, 0)
]

df_igs = df_igs[df_igs['DATETIME'].isin(dates_obj)]
```
