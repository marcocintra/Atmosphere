```python
# Download F10_7_fluxtable.txt from https://doi.org/10.5281/zenodo.15453941
```

```python
import matplotlib.pyplot as plt
import pandas as pd

def convert_to_datetime(date_str, time_str):
    return pd.to_datetime(date_str + time_str, format='%Y%m%d%H%M%S')

data = []
with open('fluxtable.txt') as file:
    next(file)
    next(file)
    for line in file:
        if line.startswith('-'):
            continue
        parts = line.split()
        if len(parts) < 7:
            continue
        flux_date = parts[0]
        flux_time = parts[1]
        flux_obs_flux = float(parts[4])
        flux_adj_flux = float(parts[5])
        datetime = convert_to_datetime(flux_date, flux_time)
        data.append((datetime, flux_obs_flux, flux_adj_flux))

df = pd.DataFrame(data, columns=['datetime', 'fluxobsflux', 'fluxadjflux'])

start_date = '2022-01-01'
end_date = '2024-12-31'
mask = (df['datetime'] >= start_date) & (df['datetime'] <= end_date)
filtered_df = df.loc[mask]
```

```python
import pandas as pd

filtered_df = filtered_df.copy()
filtered_df['datetime'] = pd.to_datetime(filtered_df['datetime'])

start_date = '2022-03-01'
end_date = '2024-12-31'
period_df = filtered_df[(filtered_df['datetime'] >= start_date) &
                         (filtered_df['datetime'] <= end_date)]
```

```python
period_df
```

```python
mask = period_df['datetime'].dt.hour.isin([22, 23])
df_22_23 = period_df[mask]

media_fluxo = df_22_23['fluxobsflux'].mean()
print(f"Average flow for 22:00 and 23:00: {media_fluxo}")
```

```python
print("Top 10 fluxobsflux values:")
top10_obs = period_df.sort_values(by='fluxobsflux', ascending=False).head(10)
print(top10_obs)

print("\nTop 10 fluxadjflux values:")
top10_adj = period_df.sort_values(by='fluxadjflux', ascending=False).head(10)
print(top10_adj)

seasonal_df = period_df[period_df['datetime'].dt.month.isin([3, 6, 9, 12])]

print("Top 10 fluxobsflux values (filtered months):")
top10_obs = seasonal_df.sort_values(by='fluxobsflux', ascending=False).head(10)
print(top10_obs)

print("\nTop 10 fluxadjflux values (filtered months):")
top10_adj = seasonal_df.sort_values(by='fluxadjflux', ascending=False).head(10)
print(top10_adj)

december_2024_df = period_df[(period_df['datetime'].dt.month == 12) &
                             (period_df['datetime'].dt.year == 2024)]

september_2024_df = period_df[(period_df['datetime'].dt.month == 9) &
                             (period_df['datetime'].dt.year == 2024)]

print("\nTop 10 fluxobsflux values (december/2024):")
top10_obs_dec = december_2024_df.sort_values(by='fluxobsflux', ascending=False).head(10)
print(top10_obs_dec)

print("\nTop 10 fluxadjflux values (december/2024):")
top10_adj_dec = december_2024_df.sort_values(by='fluxadjflux', ascending=False).head(10)
print(top10_adj_dec)

print("\nTop 10 fluxobsflux values (september/2024):")
top10_obs_sep = september_2024_df.sort_values(by='fluxobsflux', ascending=False).head(10)
print(top10_obs_sep)

print("\nTop 10 fluxadjflux values (september/2024):")
top10_adj_sep = september_2024_df.sort_values(by='fluxadjflux', ascending=False).head(10)
print(top10_adj_sep)
```
